```python
import numpy as np

# Collection of activation functions
# Reference: https://en.wikipedia.org/wiki/Activation_function
x = np.array([-2.0, 0.0, 2.0])

class Sigmoid():
    def __call__(self, x):
        return 1 / (1 + np.exp(-x))

    def gradient(self, x):
        return self.__call__(x) * (1 - self.__call__(x))

class Softmax():
    def __call__(self, x):
        e_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
        return e_x / np.sum(e_x, axis=-1, keepdims=True)

    def gradient(self, x):
        p = self.__call__(x)
        return p * (1 - p)

class TanH():
    def __call__(self, x):
        return 2 / (1 + np.exp(-2*x)) - 1

    def gradient(self, x):
        return 1 - np.power(self.__call__(x), 2)

class ReLU():
    def __call__(self, x):
        return np.where(x >= 0, x, 0)

    def gradient(self, x):
        return np.where(x >= 0, 1, 0)

class LeakyReLU():
    def __init__(self, alpha=0.2):
        self.alpha = alpha

    def __call__(self, x):
        return np.where(x >= 0, x, self.alpha * x)

    def gradient(self, x):
        return np.where(x >= 0, 1, self.alpha)

class ELU():
    def __init__(self, alpha=0.1):
        self.alpha = alpha 

    def __call__(self, x):
        return np.where(x >= 0.0, x, self.alpha * (np.exp(x) - 1))

    def gradient(self, x):
        return np.where(x >= 0.0, 1, self.__call__(x) + self.alpha)

class SELU():
    # Reference : https://arxiv.org/
    # abs/1706.02515,
    # https://github.com/bioinf-jku/SNNs/blob/master/SelfNormalizingNetworks_MLP_MNIST.ipynb
    def __init__(self):
        self.alpha = 1.6732632423543772848170429916717
        self.scale = 1.0507009873554804934193349852946 

    def __call__(self, x):
        return self.scale * np.where(x >= 0.0, x, self.alpha*(np.exp(x)-1))

    def gradient(self, x):
        return self.scale * np.where(x >= 0.0, 1, self.alpha * np.exp(x))

class SoftPlus():
    def __call__(self, x):
        return np.log(1 + np.exp(x))

    def gradient(self, x):
        return 1 / (1 + np.exp(-x))

class BinaryStep:
    def __call__(self,x):
        return np.where(x >= 0, 1, 0)
    def gradient(self,x):
        return np.zeros_like(x)

class Softsign:
    def __call__(self,x):
        return np.sign(x)*np.absolute(x)/(1+np.absolute(x))
    def gradient(self,x):
        return 1/((1+np.absolute(x))**2)

class LogofSigmoid:
    def __call__(self,x):
        return np.log(1 / (1 + np.exp(-x)))
    def gradient(self,x):
        return 1- (1 / (1 + np.exp(-x)))

class Swish:
    def __call__(self,x):
        return np.sign(x)*np.absolute(x)/(1+np.exp(-x))
    def gradient(self,x):
        return np.sign(x)*np.absolute(x)/(1+np.exp(-x)) + (1 - np.sign(x)*np.absolute(x)/(1+np.exp(-x)) )*(1 / (1 + np.exp(-x)) )

class Sinc:
    def __call__(self, x):
        return np.where(x == 0, 1.0, np.sin(x) / x)
    def gradient(self, x):
        return np.where(x == 0, 0.0, (x * np.cos(x) - np.sin(x)) / (x**2))
class Sincfix0:
    def __call__(self,x):
        return np.sin(x)/(np.sign(x)*np.absolute(x)+10**(-5))
    def gradient(self,x):
        return (np.cos(x)*np.sign(x)*np.absolute(x) - np.sin(x) )/ ((np.sign(x)*np.absolute(x)+10**(-5))**2)
class Mish:
    def __init__(self):
        self.tanh = TanH()
        self.softplus = SoftPlus()
        self.sigmoid = Sigmoid()
    def __call__(self, x):
        return x * self.tanh(self.softplus(x))
    def gradient(self, x):
        sp = self.softplus(x)
        th = self.tanh(sp)
        sig = self.sigmoid(x)
        return th + x * sig * (1 - th**2)

sigmoid=Sigmoid()
softplus=SoftPlus()
tanh=TanH()
softsign=Mish()
y_softsign= softsign(x)
y_grad= softsign.gradient(x)
print(y_softsign , y_grad)
```
